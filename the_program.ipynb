```python
import os
import tkinter as tk
from tkinter import filedialog, messagebox
import pygame

# Initialize pygame mixer
pygame.mixer.init()

def play_music(file_path):
    pygame.mixer.music.load(file_path)
    pygame.mixer.music.play()


# Ambience files ship inside the repo, in a "musics" folder next to this file
try:
    BASE_DIR = os.path.dirname(os.path.abspath(__file__))
except NameError:
    BASE_DIR = os.getcwd()
MUSICS_DIR = os.path.join(BASE_DIR, "musics")

AMBIENCE_FILES = {
    "Courage_In_The_Storm": "Courage_In_The_Storm.mp3",
    "Epic_Battle_Music": "Epic_Battle_Music.mp3",
    "Zombie_Attack": "Zombie_Attack.mp3",
    "Tavern": "Tavern.mp3",
    "The_Journey_Begins": "The_Journey_Begins.mp3",
    "Runara_Entering_First_Time": "Runara_Entering_First_Time.mp3",
    "Runara_Transforming": "Runara_Transforming.mp3",
    "Temple": "Temple_Music.mp3",
    "Kobold_Ambush": "Kobold_Ambush.mp3"
}

def play_ambience(name):
    file_name = AMBIENCE_FILES.get(name)
    if not file_name:
        return
    file_path = os.path.join(MUSICS_DIR, file_name)
    if not os.path.isfile(file_path):
        chosen = filedialog.askopenfilename(
            title=f"Locate file for {name}",
            filetypes=[("Audio files", "*.mp3 *.wav *.ogg")]
        )
        if not chosen:
            return
        file_path = chosen
    try:
        play_music(file_path)
    except pygame.error as e:
        messagebox.showerror("Playback error", f"Could not play {name}:\n{e}")


# Create the main window
root = tk.Tk()
root.title("DnD Ambience Player")


# Add ambience buttons
for ambience in AMBIENCE_FILES:
    btn = tk.Button(root, text=f"Play {ambience} Ambience", command=lambda a=ambience: play_ambience(a))
    btn.pack(pady=5)


def stop_music():
    pygame.mixer.music.stop()

stop_button = tk.Button(root, text="Stop Music", command=stop_music)
stop_button.pack(pady=10)


root.mainloop()
```
