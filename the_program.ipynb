```python
import tkinter as tk
from tkinter import filedialog
import pygame

# Initialize pygame mixer
pygame.mixer.init()

def play_music(file_path):
    pygame.mixer.music.load(file_path)
    pygame.mixer.music.play()

def choose_and_play():
    file_path = filedialog.askopenfilename(filetypes=[("Audio Files", "*.mp3 *.wav")])
    if file_path:
        play_music(file_path)


# Ambience file paths
AMBIENCE_FILES = {
    "Opening": r"C:\Users\josef\Desktop\dnd-agent\the_program\musics\opening.mp3",
    "Zombie_Attack": r"C:\Users\josef\Desktop\dnd-agent\the_program\musics\zombie_attack.mp3",
    "Tavern": r"C:\Users\josef\Desktop\dnd-agent\the_program\musics\tavern.mp3",
    "The_Journey_Begins": r"C:\Users\josef\Desktop\dnd-agent\the_program\musics\the_journey_begins.mp3",
    "Runara_Entering_First_Time": r"C:\Users\josef\Desktop\dnd-agent\the_program\musics\runara_entering_first_time.mp3",
    "Runara_Transforming": r"C:\Users\josef\Desktop\dnd-agent\the_program\musics\runara_transforming.mp3",
    "Temple_Music": r"C:\Users\josef\desktop\dnd-agent\the_program\musics\temple_music.mp3",
    "Kobold_Ambush": r"C:\Users\josef\desktop\dnd-agent\the_program\musics\kobold_ambush.mp3"
}

def play_ambience(name):
    file_path = AMBIENCE_FILES.get(name)
    if file_path:
        play_music(file_path)
    

# Create the main window
root = tk.Tk()
root.title("DnD Ambience Player")

play_button = tk.Button(root, text="Choose and Play Music", command=choose_and_play)
play_button.pack(pady=10)


# Add ambience buttons
for ambience in AMBIENCE_FILES:
    btn = tk.Button(root, text=f"Play {ambience} Ambience", command=lambda a=ambience: play_ambience(a))
    btn.pack(pady=5)


root.mainloop()
```
